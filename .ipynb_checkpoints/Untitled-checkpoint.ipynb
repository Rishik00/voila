```python
import numpy as np
from voila import evaluate, parse, vector

x = vector("x", 4)
y = vector("y", 4)

expression = (x + y) * 2
res = evaluate(
    expression, 
    {
        "x": np.array([1,2,3,4]),
        "y": np.array([2,3,4,5])
    }
)

# 4. Or parse the same expression from the supported grammar.
parsed = parse("(x + y) * 2", {"x": 4, "y": 4})
run = parsed.compile()
print("parsed result:", run([1, 2, 3, 4], [5, 6, 7, 8]).tolist())
```

```python
# print(ast.dump(ast.parse("c == d"), indent=1))

class My(ast.NodeVisitor):

    def generic_visit(self, node):
        print(f'entering {node.__class__.__name__}')
        super().generic_visit(node)
        print(f'leaving {node.__class__.__name__}')

import numpy as np

expr = """
x = np.array([1,2,3])
y = np.array([3,4,5])
z = x + y
"""

visitor = My()
tree = ast.parse(expr)
print(ast.dump(tree, indent=1))
visitor.visit(tree)
```

```python
import numpy as np
import sympy as sp
import ast
import astpretty

x_shape = 5
y_shape = 5

x, y, z = sp.symbols("x y z")

class SympyNodeVisitor(ast.NodeVisitor):
    def generic_visit(self, node):
        super().generic_visit(node)

        def node_class(a) -> str:
            return a.__class__.__name__

        print(node_class(node))


visitor = SympyNodeVisitor()

expressions = [
    # Basic arithmetic
    x + y,
    # x + 2 * y,
    # sp.sqrt(x),

    # # Elementary functions
    # sp.exp(x),
    # sp.exp(sp.sin(x) + sp.cos(y)),

    # # Rational expressions
    # 1 / (x + y),
    # (x**2 + y**2) / (x - y),

    # # Constants
    # sp.pi * x,

    # # Relational expressions
    # sp.Eq(x, y),
    # x < y,

    # # Piecewise
    # sp.Piecewise(
    #     (x**2, x < 0),
    #     (x, True),
    # ),

    # # Symbolic sum
    # sp.Sum(x**2, (x, 0, 10)),

    # # Derivatives
    # sp.Derivative(sp.sin(x), x),

    # # Integral
    # sp.Integral(sp.exp(-x**2), (x, -sp.oo, sp.oo)),
]

for i, expr in enumerate(expressions):
    print("\n" + "=" * 80)
    print(f"EXPRESSION {i}")
    print("=" * 80)

    print("\nSymPy expression:")
    print(expr)

    source = sp.srepr(expr)

    print("\nSymPy srepr:")
    print(source)

    tree = ast.parse(source)

    print("\nPython AST:")
    astpretty.pprint(
        tree,
        show_offsets=False,
        indent=2,
    )

    # print("\nVisitor traversal:")
    # visitor.visit(tree)
```

```python
import numpy as np
import sympy as sp
from functools import reduce
import ast

dtype_conv = {
    'float': np.float64
}

class ConvSymbol(sp.Symbol):
    def __new__(cls, name, shape=None, dtype=None, ):
        obj = super().__new__(cls, name)
        obj.shape = shape
        obj.dtype = dtype
        return obj

    def realize(self):
        self.data = np.ones(shape=self.shape, dtype=dtype_conv[self.dtype])
        
x = ConvSymbol('x', shape=5, dtype='float')
y = ConvSymbol('y', shape=5, dtype='float')
z = ConvSymbol('z', shape=5, dtype='float')
expr = x + y + z
tree = sp.srepr(expr)

print(tree)
# v = {}

if expr.func == sp.Add:
    for arg in expr.args:
        arg.realize()

    print(reduce(np.add, [x.data, y.data, z.data]))
```
